```python
df = spark.createDataFrame([
    (1, "rama", "bangalore"),
    (2, "krishna", "hyderabad"),
    (3, "sai", "chennai")
], ["cid", "cname", "clocation"])

display(df)
```

```python
df1 = spark.createDataFrame([
    (1, "rama", "bangalore", 8904424822),
    (2, "krishna", "chennai", 896897897),
    (3, "sai", "hyderabad", 8968969)
], ["cid", "cname", "clocation", "ccontact"])

display(df1)
```

```python
df1.coalesce(1).write.mode("overwrite").format("parquet").save("abfss://bronze@stragegentwobrcks.dfs.core.windows.net/parquetdata")
```

```python
df1.coalesce(1).write.mode("overwrite").option("mergeSchema","true").format("delta").save("abfss://bronze@stragegentwobrcks.dfs.core.windows.net/deltadata")
```
